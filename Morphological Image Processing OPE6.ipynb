```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# Load original images from OPE 6 PDF

img1 = cv2.imread("img-000.png")
img2 = cv2.imread("img-001.jpg")
img3 = cv2.imread("img-002.png")

# Convert images to RGB for display
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
img3 = cv2.cvtColor(img3, cv2.COLOR_BGR2RGB)

# Display all three images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img1)
plt.title("Binary Object")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(img2)
plt.title("Old Photograph")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(img3)
plt.title("Fingerprint")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Convert RGB images to grayscale
gray1 = cv2.cvtColor(img1, cv2.COLOR_RGB2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_RGB2GRAY)
gray3 = cv2.cvtColor(img3, cv2.COLOR_RGB2GRAY)

# Convert grayscale images to binary
_, binary1 = cv2.threshold(
    gray1, 127, 255, cv2.THRESH_BINARY
)

_, binary2 = cv2.threshold(
    gray2, 127, 255, cv2.THRESH_BINARY
)

_, binary3 = cv2.threshold(
    gray3, 127, 255, cv2.THRESH_BINARY
)

# Display binary images
plt.figure(figsize=(15, 6))

plt.subplot(1, 3, 1)
plt.imshow(binary1, cmap='gray')
plt.title("Binary Image 1")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(binary2, cmap='gray')
plt.title("Binary Image 2")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(binary3, cmap='gray')
plt.title("Binary Image 3")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Create structuring element
kernel = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE, (3, 3)
)

# Apply morphological opening
opening1 = cv2.morphologyEx(
    binary1, cv2.MORPH_OPEN, kernel
)

opening2 = cv2.morphologyEx(
    binary2, cv2.MORPH_OPEN, kernel
)

opening3 = cv2.morphologyEx(
    binary3, cv2.MORPH_OPEN, kernel
)

# Display results
plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.imshow(binary1, cmap='gray')
plt.title("Original Image 1")
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(binary2, cmap='gray')
plt.title("Original Image 2")
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(binary3, cmap='gray')
plt.title("Original Image 3")
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(opening1, cmap='gray')
plt.title("Opening Image 1")
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(opening2, cmap='gray')
plt.title("Opening Image 2")
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(opening3, cmap='gray')
plt.title("Opening Image 3")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Create structuring element for closing
kernel_close = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE, (5, 5)
)

# Apply morphological closing
closing1 = cv2.morphologyEx(
    opening1, cv2.MORPH_CLOSE, kernel_close
)

closing2 = cv2.morphologyEx(
    opening2, cv2.MORPH_CLOSE, kernel_close
)

closing3 = cv2.morphologyEx(
    opening3, cv2.MORPH_CLOSE, kernel_close
)

# Display final results
plt.figure(figsize=(15, 8))

plt.subplot(1, 3, 1)
plt.imshow(closing1, cmap='gray')
plt.title("Final Image 1")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(closing2, cmap='gray')
plt.title("Final Image 2")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(closing3, cmap='gray')
plt.title("Final Image 3")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 12))

# Image 1
plt.subplot(3, 3, 1)
plt.imshow(binary1, cmap='gray')
plt.title("Image 1 - Original")
plt.axis('off')

plt.subplot(3, 3, 2)
plt.imshow(opening1, cmap='gray')
plt.title("Image 1 - Opening")
plt.axis('off')

plt.subplot(3, 3, 3)
plt.imshow(closing1, cmap='gray')
plt.title("Image 1 - Final")
plt.axis('off')

# Image 2
plt.subplot(3, 3, 4)
plt.imshow(binary2, cmap='gray')
plt.title("Image 2 - Original")
plt.axis('off')

plt.subplot(3, 3, 5)
plt.imshow(opening2, cmap='gray')
plt.title("Image 2 - Opening")
plt.axis('off')

plt.subplot(3, 3, 6)
plt.imshow(closing2, cmap='gray')
plt.title("Image 2 - Final")
plt.axis('off')

# Image 3
plt.subplot(3, 3, 7)
plt.imshow(binary3, cmap='gray')
plt.title("Image 3 - Original")
plt.axis('off')

plt.subplot(3, 3, 8)
plt.imshow(opening3, cmap='gray')
plt.title("Image 3 - Opening")
plt.axis('off')

plt.subplot(3, 3, 9)
plt.imshow(closing3, cmap='gray')
plt.title("Image 3 - Final")
plt.axis('off')

plt.tight_layout()
plt.show()
```
