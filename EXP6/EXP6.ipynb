```python
#1. Design FIR Low pass filter having cutoff frequency 3 KHz, N=31; Sampling frequency=10 KHz
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import firwin, freqz

# Step 1: Define filter parameters
Fs = 10000       
Fc = 3000        
N = 31           

# Step 2: Design FIR low-pass filter using Hann window
b = firwin(
    numtaps=N,
    cutoff=Fc,
    window='hann',
    pass_zero='lowpass',
    fs=Fs
)
f, H = freqz(b, worN=2048, fs=Fs)

magnitude_db = 20 * np.log10(np.maximum(np.abs(H), 1e-10))
phase = np.unwrap(np.angle(H))

# Plot impulse, magnitude and phase responses
plt.figure(figsize=(12, 10))

plt.subplot(3, 1, 1)
plt.stem(np.arange(N), b)
plt.title('FIR Low-Pass Filter - Impulse Response')
plt.xlabel('Sample n')
plt.ylabel('Amplitude')
plt.grid(True)

plt.subplot(3, 1, 2)
plt.plot(f, magnitude_db)
plt.axvline(Fc, color='red', linestyle='--',
            label='Cutoff = 3 kHz')
plt.title('Magnitude Response')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude (dB)')
plt.xlim(0, Fs / 2)
plt.legend()
plt.grid(True)

plt.subplot(3, 1, 3)
plt.plot(f, phase)
plt.title('Phase Response')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Phase (radians)')
plt.xlim(0, Fs / 2)
plt.grid(True)

plt.tight_layout()
plt.show()

```

```python
# FIR Low-Pass Filter Design Using Hann Window with Music
# Cutoff Frequency = 3 kHz, N = 31

import numpy as np
import matplotlib.pyplot as plt
import soundfile as sf
import librosa
from scipy.signal import firwin, freqz, lfilter
from IPython.display import Audio, display

# Step 1: Load MP3 audio
audio_path = r"c:\Users\shash\Music\Garba.mp3"

music, Fs = librosa.load(audio_path, sr=None, mono=True)

print("Sampling Frequency:", Fs, "Hz")
print("Audio Duration:", len(music) / Fs, "seconds")

# Step 2: Define filter parameters
Fc = 3000
N = 31

if Fc >= Fs / 2:
    raise ValueError("Cutoff frequency must be below Nyquist frequency.")

# Step 3: Design FIR low-pass filter using Hann window
b = firwin(
    numtaps=N,
    cutoff=Fc,
    window='hann',
    pass_zero='lowpass',
    fs=Fs
)

# Step 4: Filter the music
filtered_music = lfilter(b, [1.0], music)

# Step 5: Calculate frequency response
f, H = freqz(b, worN=2048, fs=Fs)

magnitude_db = 20 * np.log10(
    np.maximum(np.abs(H), 1e-10)
)
phase = np.unwrap(np.angle(H))

# Step 6: Plot filter responses
plt.figure(figsize=(12, 10))

plt.subplot(3, 1, 1)
plt.stem(np.arange(N), b)
plt.title("FIR Low-Pass Filter - Impulse Response")
plt.xlabel("Sample n")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(3, 1, 2)
plt.plot(f, magnitude_db)
plt.axvline(Fc, color='red', linestyle='--',
            label='Cutoff = 3 kHz')
plt.title("Magnitude Response")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude (dB)")
plt.legend()
plt.grid(True)

plt.subplot(3, 1, 3)
plt.plot(f, phase)
plt.title("Phase Response")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Phase (radians)")
plt.grid(True)

plt.tight_layout()
plt.show()

# Step 7: Plot first 5 seconds of original and filtered music
samples = min(5 * Fs, len(music))
time = np.arange(samples) / Fs

plt.figure(figsize=(12, 6))

plt.subplot(2, 1, 1)
plt.plot(time, music[:samples])
plt.title("Original Music Signal")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(time, filtered_music[:samples])
plt.title("Filtered Music Signal - 3 kHz Low-Pass")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()

# Step 8: Listen to original music
print("Original Music")
display(Audio(music, rate=Fs))

# Step 9: Listen to filtered music
print("Filtered Music")
display(Audio(filtered_music, rate=Fs))

# Step 10: Save filtered audio
sf.write("filtered_music.wav", filtered_music, Fs)

print("Filtered audio saved as filtered_music.wav")
```
