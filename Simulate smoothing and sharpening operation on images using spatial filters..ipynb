```python
#1Smoothing and Sharpening

import cv2
import numpy as np


image = cv2.imread(r"c:\Users\shash\Documents\SHASHIDAHR_18\Photos\5.jpg")

if image is None:
    print("Error: Image not found!")
    exit()

kernel_size = (5, 5)
sigma = 1.5

gaussian_kernel = cv2.getGaussianKernel(kernel_size[0], sigma)
gaussian_kernel = np.outer(gaussian_kernel, gaussian_kernel)

smoothed_image = cv2.filter2D(image, -1, gaussian_kernel)

sharpening_kernel = np.array([
    [-1, -1, -1],
    [-1,  9, -1],
    [-1, -1, -1]
])

sharpened_image = cv2.filter2D(image, -1, sharpening_kernel)

# Display images
cv2.imshow("Original Image", image)
cv2.imshow("Smoothed Image", smoothed_image)
cv2.imshow("Sharpened Image", sharpened_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#2Smoothing using Averaging Linear Filter

import cv2
import numpy as np

# Load the image
image = cv2.imread(
    r"c:\Users\shash\Documents\SHASHIDAHR_18\Photos\hello_51.jpg"
)

# Check if image was loaded successfully
if image is None:
    print("Error: Image not found!")
    exit()

# Define the size of the averaging filter kernel
kernel_size = (5, 5)

# Create the Averaging filter kernel
kernel = np.ones(
    kernel_size,
    dtype=np.float32
) / (kernel_size[0] * kernel_size[1])

# Apply the Averaging filter for smoothing
smoothed_image = cv2.filter2D(image, -1, kernel)

# Display the original and smoothed images
cv2.imshow("Original Image", image)
cv2.imshow("Averaging Filter - Smoothed Image", smoothed_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#3Smoothing using Median Filter

import cv2
import numpy as np

# Load the image
image = cv2.imread(
    r"c:\Users\shash\Documents\SHASHIDAHR_18\Photos\hello_9.jpg"
)

# Check if image was loaded successfully
if image is None:
    print("Error: Image not found!")
    exit()

# Define the size of the median filter kernel
# Kernel size should be an odd number
kernel_size = 5

# Apply the Median filter for smoothing
smoothed_image = cv2.medianBlur(image, kernel_size)

# Display the original and smoothed images
cv2.imshow("Original Image", image)
cv2.imshow("Median Filter - Smoothed Image", smoothed_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#4Something and Sharpening Using Spatial High-Pass Filters

import cv2
import numpy as np

# Load the image
image = cv2.imread(r'c:\Users\shash\Documents\SHASHIDAHR_18\Photos\KA group pic.jpg')

# Check if image is loaded
if image is None:
    print("Error: Image not found!")
    exit()

# Apply Gaussian smoothing to reduce noise
blurred_image = cv2.GaussianBlur(image, (5, 5), 0)

# Create a spatial high-pass sharpening kernel
laplacian_kernel = np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
], dtype=np.float32)

# Apply high-pass filter
sharpened_image = cv2.filter2D(
    blurred_image,
    -1,
    laplacian_kernel
)

# Display images
cv2.imshow("Original Image", image)
cv2.imshow("Blurred Image", blurred_image)
cv2.imshow("Sharpened Image", sharpened_image)

# Wait for key press
cv2.waitKey(0)

# Close all windows
cv2.destroyAllWindows()
```
